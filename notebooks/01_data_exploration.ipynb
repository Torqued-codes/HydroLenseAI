# 01 — Data Exploration
Explore the raw water-quality data and visualize distributions.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_STATE = 42
ROOT = Path.cwd().parent
RAW = ROOT/"data"/"raw"/"water_quality_raw.csv"
PROC = ROOT/"data"/"processed"
MODELS = ROOT/"models"

if RAW.exists():
    df=pd.read_csv(RAW)
else:
    rng=np.random.default_rng(RANDOM_STATE); n=1000
    df=pd.DataFrame({
      "date":pd.date_range("2026-01-01",periods=n,freq="D"),
      "location":rng.choice(["Location_A","Location_B","Location_C"],n),
      "pH":rng.normal(7.2,.25,n),"turbidity":np.clip(rng.normal(2.5,.8,n),.1,None),
      "tds":np.clip(rng.normal(350,80,n),50,None),"temperature":rng.normal(25,2.5,n),
      "dissolved_oxygen":np.clip(rng.normal(7,.9,n),1,None),
      "conductivity":np.clip(rng.normal(550,120,n),50,None)})
display(df.head()); print("Shape:",df.shape)
```

```python
print(df.isna().sum()); display(df.select_dtypes("number").describe().T)
```

```python
df.select_dtypes("number").hist(figsize=(12,8),bins=30); plt.tight_layout(); plt.show()
```

```python
PROC.mkdir(parents=True,exist_ok=True); df.to_csv(PROC/"exploration_snapshot.csv",index=False)
```
