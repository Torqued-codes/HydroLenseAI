# 05 — Demo Predictions
Run reproducible normal and intentionally unusual demo samples.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RANDOM_STATE = 42
ROOT = Path.cwd().parent
RAW = ROOT/"data"/"raw"/"water_quality_raw.csv"
PROC = ROOT/"data"/"processed"
MODELS = ROOT/"models"

import joblib
model=joblib.load(MODELS/"anomaly_detector.pkl"); scaler=joblib.load(MODELS/"scaler.pkl")
meta=json.loads((MODELS/"model_metadata.json").read_text()); features=meta["features"]
normal_sample=pd.DataFrame([{"pH":7.2,"turbidity":2.4,"tds":340,"temperature":25.0,"dissolved_oxygen":7.1,"conductivity":540}])[features]
anomaly_sample=pd.DataFrame([{"pH":5.4,"turbidity":20.0,"tds":1200,"temperature":31.0,"dissolved_oxygen":2.0,"conductivity":1800}])[features]
```

```python
def analyze(sample):
    x=scaler.transform(sample); p=int(model.predict(x)[0]); s=float(model.decision_function(x)[0])
    return {"label":"ANOMALY DETECTED" if p==-1 else "NO SIGNIFICANT ANOMALY DETECTED","decision_score":s}
print("Normal:",analyze(normal_sample)); print("Unusual:",analyze(anomaly_sample))
```

```python
print("Interpretation: the model flags unusual patterns relative to training data. It does not certify drinking-water safety; real decisions require appropriate verification.")
```
